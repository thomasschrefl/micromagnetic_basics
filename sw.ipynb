# Small particles

### anisotropy energy density and Zeeman energy
 
$E_\mathrm{ani} =      K \sin^2(\varphi) V$

$E_\mathrm{zee} = - H J_\mathrm{s} \cos(\theta-\varphi) V$, $J_\mathrm{s} = \mu_0 M_\mathrm{s}$

### normalize energies: divide by $2KV$

$e_\mathrm{ani} =      \frac{1}{2} \sin^2(\varphi)$

$e_\mathrm{zee} = - h \cos(\theta-\varphi)$, with $h = H \frac{J_\mathrm{s}}{2K}$ 

```python
import matplotlib
import matplotlib.pyplot as plt
matplotlib.rcParams.update({'font.size': 12})

import numpy as np
```

### energy density

```python
def energy_density(phi, h, theta):
  e_ani = 0.5*(np.sin(phi))**2
  e_zee = - h*np.cos(theta-phi)
  return (e_ani + e_zee).squeeze()
```

```python
theta    = np.radians(0.0)       # field angle
h_values = np.linspace(0,-1,5)   # field values

# magnetization angles for plotting
phi = np.linspace(np.radians(0-10),np.radians(180+10),100) 

for h in h_values:

  plt.plot(np.degrees(phi),energy_density(phi,h,theta))

  plt.title(f'normalized field: {h}, field angle: {np.degrees(theta)}') 
  plt.xlabel('magnetization angle')
  plt.ylabel('energy density ($2KV$)')
  plt.xlim((-10.,190.0))
  plt.xticks(np.linspace(0,180,10))
  plt.ylim((-1.05,1.05))
  plt.show()
```
